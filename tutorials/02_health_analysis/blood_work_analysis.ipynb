```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_groq import ChatGroq

load_dotenv()
llm_genai=ChatGoogleGenerativeAI(model="gemmma-4-31b-it")
llm_groq = ChatGroq(model="qwen/qwen3.6-27b",temperature=0,reasoning_effort="none")  # we can use reasoning_format too => parsed, raw, hidden
```

```python
with open("blood_work.txt","r") as f:
    blood_report=f.read()
print(blood_report[:200])
```
